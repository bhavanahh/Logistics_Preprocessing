```python
delayed_count = df["delayed_yes"].sum()
not_delayed_count = len(df) - delayed_count

print("Delayed Deliveries:", delayed_count)
print("On-Time Deliveries:", not_delayed_count)
```

```python
delay_data = pd.Series({
    "On Time": not_delayed_count,
    "Delayed": delayed_count
})

plt.figure(figsize=(7,5))

delay_data.plot(kind="bar")

plt.title("Delayed vs On-Time Deliveries")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Deliveries")
plt.xticks(rotation=0)

plt.show()
```

```python
weather_delay_rates = {}

for weather in weather_columns:
    condition = df[weather] == True
    delay_rate = df.loc[condition, "delayed_yes"].mean()
    weather_delay_rates[weather] = delay_rate

weather_delay_rates = pd.Series(weather_delay_rates)

print(weather_delay_rates)
```

```python
weather_delay_percentage = weather_delay_rates * 100

print(weather_delay_percentage)
```

```python
plt.figure(figsize=(9,5))

weather_delay_percentage.plot(kind="bar")

plt.title("Delivery Delay Rate by Weather Condition")
plt.xlabel("Weather Condition")
plt.ylabel("Delay Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

```python
highest_delay_weather = weather_delay_percentage.idxmax()
highest_delay_rate = weather_delay_percentage.max()

print("Highest Delay Weather Condition:", highest_delay_weather)
print("Delay Rate:", round(highest_delay_rate, 2), "%")
```

```python
lowest_delay_weather = weather_delay_percentage.idxmin()
lowest_delay_rate = weather_delay_percentage.min()

print("Lowest Delay Weather Condition:", lowest_delay_weather)
print("Delay Rate:", round(lowest_delay_rate, 2), "%")
```

```python
print(weather_delay_percentage)
```

Interpretation: The analysis shows that weather conditions have an important relationship with delivery delays. Stormy weather recorded the highest delay rate at 41.45%, followed by rainy weather at 37.35%. Foggy conditions resulted in a delay rate of 30.32%, while hot and cold conditions recorded comparatively lower delay rates of 17.12% and 16.02%, respectively. These findings suggest that adverse weather conditions, particularly storms and heavy rain, may create significant operational bottlenecks in logistics delivery.

```python
late_deliveries = (
    df["delivery_time_hours"] >
    df["expected_time_hours"]
).sum()

on_time_deliveries = len(df) - late_deliveries

print("Late Deliveries:", late_deliveries)
print("On-Time/Early Deliveries:", on_time_deliveries)
```

```python
late_percentage = (late_deliveries / len(df)) * 100

print(
    "Percentage of Late Deliveries:",
    round(late_percentage, 2),
    "%"
)
```

Interpretation: The comparison between actual and expected delivery times shows an average difference of -0.18 hours, indicating that deliveries were completed approximately 11 minutes earlier than the expected time on average. However, 9,915 deliveries, representing 39.66% of the total deliveries, took longer than their expected delivery time. This indicates that while overall delivery performance was slightly better than expected, a considerable proportion of shipments experienced delays. Therefore, logistics managers should investigate the factors contributing to individual delivery delays rather than relying solely on average delivery performance.

```python
numeric_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_time_hours",
    "expected_time_hours",
    "delivery_rating",
    "delivery_cost"
]

correlation_matrix = df[numeric_columns].corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Logistics Variables")

plt.show()
```

Interpretation: The correlation analysis reveals several important relationships among logistics performance variables. Distance and delivery cost have the strongest positive correlation, with a coefficient of 0.991, indicating that transportation distance is a major factor associated with delivery cost. Distance also has a positive correlation of 0.686 with delivery time, suggesting that longer-distance deliveries generally require more time. Delivery time and delivery cost show a correlation of 0.679, indicating that longer delivery durations are also associated with increased delivery costs. In contrast, package weight has only a weak correlation with delivery cost (0.099), suggesting that weight is not a major cost driver in this dataset. Delivery rating has a negative correlation of -0.273 with delivery time, indicating that longer delivery times are associated with somewhat lower customer ratings. These relationships provide useful evidence for identifying cost drivers and areas for operational improvement.

```python
partner_performance = {}

for partner in partner_columns:
    partner_performance[partner] = {
        "Average Delivery Time": df.loc[
            df[partner] == True,
            "delivery_time_hours"
        ].mean(),
        "Average Delivery Cost": df.loc[
            df[partner] == True,
            "delivery_cost"
        ].mean(),
        "Average Rating": df.loc[
            df[partner] == True,
            "delivery_rating"
        ].mean()
    }

partner_performance_df = pd.DataFrame(partner_performance).T

print(partner_performance_df)
```

```python
partner_performance_df = partner_performance_df.round(2)

print(partner_performance_df)
```

```python
plt.figure(figsize=(10,5))

partner_performance_df["Average Rating"].plot(kind="bar")

plt.title("Average Customer Rating by Delivery Partner")
plt.xlabel("Delivery Partner")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)

plt.show()
```

Therefore, delivery partner selection can be optimized according to operational priorities, with Delhivery being favorable for time and cost efficiency and Blue Dart showing stronger performance in customer ratings.
